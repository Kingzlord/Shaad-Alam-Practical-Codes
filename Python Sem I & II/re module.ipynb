                                                        RE MODULE

```python
# Spliting
import re
s = input("Enter any string: ")
r_s1 = re.split("\\s",s)
print("The splited string is:",r_s1)
```

```python
r_s2 = re.split("[a-d]",s)
print("The splited string is:",r_s2)
```

```python
r_s3 = re.findall("\\d",s)
print("\nNumbers Splited:",r_s3)
r_s4 = re.findall("\\D",s)
print("\nOnly Strings :", r_s4)
```

```python
r_s5 = re.findall("\\s",s)
print("The splited string is:", r_s5)
# r_s6 = re.findall("\\S",s)
# print("The splited string is:",r_s6)
```

```python
r_s7 = re.findall("\\w",s)
print("The splited string is:",r_s7)
r_s8 = re.findall("\\W",s)
print("The splited string is:",r_s8)
```

```python
r_s9 = re.findall("\\bShaad",s)
print("The splited string is:",r_s9)
```

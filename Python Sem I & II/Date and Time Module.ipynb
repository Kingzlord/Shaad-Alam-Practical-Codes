Date and Time Module

```python
import datetime as dt
# Week day short version/ abbreviated
dt.datetime.now().strftime("%a")
```

```python
# Week day full version/ non-abbreviated
dt.datetime.now().strftime("%A")
```

```python
# Number coresspodning to the week day
dt.datetime.now().strftime("%w")
```

```python
# Week no. of the year
dt.datetime.now().strftime("%W")
```

```python
# Day of the month
dt.datetime.now().strftime("%d")
```

```python
# Full format
dt.datetime.now().strftime("%D")
```

```python
# Month name abbreviated
dt.datetime.now().strftime("%b")
```

```python
# 
dt.datetime.now().strftime("%B")
```

```python
#No. of the month
dt.datetime.now().strftime("%m")
```

```python
# No. of year in fullform
dt.datetime.now().strftime("%Y")
```
